```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# read the csv file
order = pd.read_csv('orders.csv')
order.head()
```

```python
# lets check the datatype
order.info()
```

```python
# lets see how many null values in each column
order.isnull().sum()
```

```python
order.columns = order.columns.str.lower() # this will make the column names to lower case
```

```python
# lets cross verify it 
order.head()
```

```python
order.columns
```

```python
# now lets check the data for each column
order['order id'].nunique() # its a pk column
```

```python
# lets check order_date
order['order date'].dtype # so its object, we need to convert it to datetime
```

```python
order['order date'] = pd.to_datetime(order['order date'])
```

```python
order['order date'] # now its ok
```

```python
# lets check ship mode
order['ship mode'].unique()
```

```python
# now convert the Not Available and unknow to nan
order['ship mode'] = order['ship mode'].replace(['Not Available','unknown'],np.nan)
```

```python
order['ship mode'].unique() # so we have replaced unknown and Not Available by nan
```

```python
# also one more thing column names have space , i want to replace by _
order.columns = order.columns.str.replace(' ','_')
order.columns
```

```python
order.head() # its now ok and easy to access
```

```python
# now lets see the segment
order['segment'].unique()
```

```python
print(order['country'].unique())# now lets see the country
print(order['city'].nunique())# now lets see how many unique cities
print(order['state'].nunique())# now lets see how many unique cities
```

```python
# lets see the postal code column
print(order['postal_code'].nunique())
print(order['postal_code'].dtype)
```

```python
print(order['region'].unique())
print(order['category'].unique())
print(order['sub_category'].unique())
```

```python
# overall rest columns dtype looks ok and we can proceed further
# so we have cost_price, list_price, quantity, discount_percent

# lets add discount_amount =  list_price * (discount/100)
# lets add net_sale_price = list_price - discount_amount

order['discount_amount'] = order['list_price']*(order['discount_percent']/100)
order['net_sale_price'] = order['list_price'] - order['discount_amount']
```

```python
order.head()
```

```python
# add a profit column
order['profit'] = order['net_sale_price'] - order['cost_price']
```

```python
order.head()
```

```python
# lets drop some columns which are not required
order.head()
# so we dont need discount_percent as we have amount, so lets drop that only
```

```python
order.drop(columns = 'discount_percent', inplace=True)
```

```python
order.head() # check
```

```python
# now lets import again
import sqlalchemy
import pymysql
from sqlalchemy import create_engine, text
```

```python
print('pymysql :',pymysql.__version__) # ok now its loaded 
```

```python
# lets load data the data .env data and connect 
import os
from dotenv import load_dotenv
env = load_dotenv('.env') # this will load .env file
host = os.getenv("host")
username = os.getenv("username")
password = os.getenv("password")
database = os.getenv("database")
port = os.getenv("port")

try:
    engine = create_engine(
    f"mysql+pymysql://{username}:{password}@{host}:{port}/{database}"
    )
    with engine.connect() as connection:
        result = connection.execute(text("select 1 as 'row', 2 as 'col';"))
        row = result.fetchall()
        print(row)
        print('Connected succesfully')
except Exception as e:
    print(f"Connection issue :{e}")
```

```python
with engine.connect() as connection:
    order.to_sql("order",con = connection,index=False, if_exists='replace')
```
